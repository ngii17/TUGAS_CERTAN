```python
import pandas as pd
path = r"D:\semester5\CERTAN1\archive\shopping_trends.csv"
df = pd.read_csv(path)
df
```

```python
mean_val = df['Previous Purchases'].mean()
print("Mean:", mean_val)
```

## Tugas 1a
 Mean
Dengan melihat data mean dapat disimpulkan bahwa pelanggan sudah pernah bebelanja selumnya dengan rata rata penah belanja itu 25 kali

```python
std_val = df['Previous Purchases'].std()
print("Std:", std_val)
```

## Tugas 1b
 std
 seberapa bevariasi kebiasaan pelanggan itu bedasarkan data cukup besar dimana bedasarkan data itu 14 koma sekian sehingga mengartikan ada yang jarang belanja dan ada yang sering belanja

```python
min_val = df['Previous Purchases'].min()
print("Min:", min_val)
```

## Tugas 1c
Min
berdasarkan data dari min yaitu 1 berarti ada pelanggan yang hanya masih sekali yang belanja

```python
max_val = df['Previous Purchases'].max()
print("Max:", max_val)
```

## Tugas 1d
Max
Berdasarkan data itu pelanggan yanng paling banyak belanja itu sebanyak 50 kali

```python
q1_val = df['Previous Purchases'].quantile(0.25)
print("Q1:", q1_val)
```

## Tugas 1g
Q3
berdasarkan data 75 dari 100 pelanggan sudah pernah belanja 38 kali ke bawah  jadi 25 dari 100 lah yang paling sering berlanja yaitu lebih dari 38 kali

## Tugas 2a.

berdasarkan data maka untuk Nilai korelasi antara Age dan Purchase Amount (USD) adalah -0.0104 yang sangat mendekati adalah 0 sehingga menunjukkan hampir tidak ada hubungan antara umur pelanggan dengan jumlah uang yang mereka belanjakan dimana dengan ini artinya umur tidak bisa dijadikan patokan untuk memprediksi besarnya nominal belanja seseorang, karena pelanggan tua maupun muda menunjukkan pola belanja yang relatif sama
